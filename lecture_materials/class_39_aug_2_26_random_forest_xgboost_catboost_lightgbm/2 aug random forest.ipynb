data = [
    (101, "Rahul", 80000, "India", "Fraud"),
    (102, "Amit", 30000, "India", "Safe"),
    (103, "John", 90000, "USA", "Fraud"),
    (101, "Rahul", 80000, "India", "Fraud"),
    (104, "Sara", 40000, "USA", "Safe"),
    (105, "Mike", 120000, "USA", "Fraud")
]

1. Remove duplicate records
2. Convert data into dictionary format
3. create a set of unique countries
4. identify high income customers (>80k)
5. count fraud customers
6. skip customers with income < 40k
7. stop processing if income > 1 lakh found
8. store final result as tuple

```python
data = [
    (101, "Rahul", 80000, "India", "Fraud"),
    (102, "Amit", 30000, "India", "Safe"),
    (103, "John", 90000, "USA", "Fraud"),
    (101, "Rahul", 80000, "India", "Fraud"),
    (104, "Sara", 40000, "USA", "Safe"),
    (105, "Mike", 120000, "USA", "Fraud")
]

unique_data = list(set(data))

customers = []
countries = set()
fraud_count = 0
results = []

for row in unique_data:
    id, name, income, country , status = row
    if income < 40000:
        continue
    if income > 100000:
        print("Very high value customer")
        break
    customer = {"id": id, "name": name, "country": country, "status": status}
    customers.append(customer)
    countries.add(country)
    if status == "Fraud":
        fraud_count += 1
    if income > 80000:
        if status == "Fraud":
            label = "High risk and high value"
        else:
            label = "High value safe"
    else:
        label = "Normal"
    output = (id, label)
    results.append(output)

print("Customers: ", customers)
print("Countries: ", countries)
print("Fraud Count: ", fraud_count)
print("Results: ", results)
```

```python
import numpy as np

# ─────────────────────────────────────────────────────────────
# RECOMMENDATION SYSTEM USING SVD (Singular Value Decomposition)
# ─────────────────────────────────────────────────────────────

# Movies list
movies = ['Inception', 'Avengers', 'Titanic', 'Matrix', 'Notebook']

# Users list
users = ['Alice', 'Bob', 'Carol', 'Dave', 'Eve']

# User-Movie Ratings Matrix
# Rows    = Users
# Columns = Movies
#
# 0 means user has NOT watched/rated that movie

ratings = np.array([
    [5, 3, 0, 1, 0],   # Alice
    [4, 0, 4, 1, 2],   # Bob
    [1, 1, 0, 5, 0],   # Carol
    [0, 0, 4, 4, 3],   # Dave
    [0, 2, 0, 0, 5]    # Eve
], dtype=float)

# ─────────────────────────────────────────────────────────────
# PRINT ORIGINAL MATRIX
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("ORIGINAL USER-MOVIE RATINGS MATRIX")
print("=" * 70)

# Print movie names as column headers
print(f"{'User':10}", end="")

for movie in movies:
    print(f"{movie:12}", end="")

print()

# Print ratings row by row
for i, user in enumerate(users):

    print(f"{user:10}", end="")

    for rating in ratings[i]:
        print(f"{rating:<12.0f}", end="")

    print()

print()

# ─────────────────────────────────────────────────────────────
# APPLY SVD
# ─────────────────────────────────────────────────────────────

# SVD breaks matrix into:
#
# A = U × Sigma × Vᵀ
#
# U     → User features
# Sigma → Importance / strength
# Vᵀ    → Movie features

U, sigma, Vt = np.linalg.svd(ratings, full_matrices=False)

# ─────────────────────────────────────────────────────────────
# PRINT U MATRIX
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("U MATRIX (USER FEATURES)")
print("=" * 70)

print(U)
print()

# ─────────────────────────────────────────────────────────────
# PRINT SIGMA VALUES
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("SIGMA VALUES (IMPORTANCE OF LATENT FACTORS)")
print("=" * 70)

print(sigma)
print()

# ─────────────────────────────────────────────────────────────
# PRINT V TRANSPOSE MATRIX
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("Vᵀ MATRIX (MOVIE FEATURES)")
print("=" * 70)

print(Vt)
print()

# ─────────────────────────────────────────────────────────────
# REDUCE DIMENSIONS
# ─────────────────────────────────────────────────────────────

# Keep only top 2 hidden factors
#
# Example hidden factors:
# - Action movie lover
# - Romantic movie lover

k = 2

# Create diagonal sigma matrix using top k sigma values
sigma_k = np.diag(sigma[:k])

# Reduce matrices
U_k = U[:, :k]

Vt_k = Vt[:k, :]

# ─────────────────────────────────────────────────────────────
# RECONSTRUCT / PREDICT MISSING RATINGS
# ─────────────────────────────────────────────────────────────

# Formula:
#
# Predicted Matrix =
# U_k × Sigma_k × Vt_k

predicted = U_k @ sigma_k @ Vt_k

# ─────────────────────────────────────────────────────────────
# PRINT PREDICTED RATINGS MATRIX
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print(f"PREDICTED RATINGS MATRIX USING SVD (k = {k})")
print("=" * 70)

# Print movie names
print(f"{'User':10}", end="")

for movie in movies:
    print(f"{movie:12}", end="")

print()

# Print predicted ratings
for i, user in enumerate(users):

    print(f"{user:10}", end="")

    for rating in predicted[i]:
        print(f"{rating:<12.2f}", end="")

    print()

print()

# ─────────────────────────────────────────────────────────────
# RECOMMEND MOVIES FOR ALICE
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("MOVIE RECOMMENDATIONS FOR ALICE")
print("=" * 70)

# ratings[0] means Alice's original ratings

for j, movie in enumerate(movies):

    # If Alice has NOT rated the movie
    if ratings[0, j] == 0:

        print(f"{movie:12} ---> Predicted Rating = {predicted[0, j]:.2f}")

print()

# ─────────────────────────────────────────────────────────────
# BEST RECOMMENDATION FOR ALICE
# ─────────────────────────────────────────────────────────────

best_movie = None
best_score = -1

for j, movie in enumerate(movies):

    # Only unseen movies
    if ratings[0, j] == 0:

        # Check predicted score
        if predicted[0, j] > best_score:

            best_score = predicted[0, j]
            best_movie = movie

print("=" * 70)
print("BEST RECOMMENDATION FOR ALICE")
print("=" * 70)

print(f"{best_movie} with predicted rating {best_score:.2f}")

print()

# ─────────────────────────────────────────────────────────────
# EXPLANATION
# ─────────────────────────────────────────────────────────────

print("=" * 70)
print("WHAT SVD IS DOING")
print("=" * 70)

print("""
1. SVD finds hidden patterns in users and movies.

2. It discovers hidden features like:
   - Action movie lovers
   - Romantic movie lovers
   - Sci-fi lovers

3. It compresses the matrix into important information only.

4. Then it reconstructs missing ratings.

5. Those reconstructed ratings become recommendations.
""")
```

```python
# ==========================================
# IMPORT LIBRARIES
# ==========================================

# Binomial distribution object
from scipy.stats import binom

# NumPy for numerical operations
import numpy as np

# ==========================================
# BUSINESS INPUTS
# ==========================================

# Total customers receiving email
n_customers = 1000

# Historical click probability
click_probability = 0.10

# Exact clicks we want to analyze
target_clicks = 120

# ==========================================
# EXPECTED VALUE
# ==========================================

# Mean number of clicks
expected_clicks = binom.mean(
    n=n_customers,
    p=click_probability
)

# ==========================================
# VARIANCE
# ==========================================

variance_clicks = binom.var(
    n=n_customers,
    p=click_probability
)

# ==========================================
# EXACT PROBABILITY
# ==========================================

exact_probability = binom.pmf(
    k=target_clicks,
    n=n_customers,
    p=click_probability
)

# ==========================================
# CUMULATIVE PROBABILITY
# ==========================================

cumulative_probability = binom.cdf(
    k=target_clicks,
    n=n_customers,
    p=click_probability
)

# ==========================================
# RANDOM SIMULATION
# ==========================================

simulated_campaign = binom.rvs(
    n=n_customers,
    p=click_probability,
    size=10
)

# ==========================================
# PRINT RESULTS
# ==========================================

print("Expected Clicks:")
print(expected_clicks)

print("\nVariance:")
print(variance_clicks)

print("\nProbability Exactly 120 Clicks:")
print(exact_probability)

print("\nProbability 120 Or Fewer Clicks:")
print(cumulative_probability)

print("\n10 Simulated Campaign Results:")
print(simulated_campaign)
```

```python
import numpy as np
import matplotlib.pyplot as plt

experience = np.array([1,2,3,4,5])

salary = np.array([3,5,7,9,11])

beta0 = 0

beta1 = 0

learning_rate = 0.01

epochs = 2000

n = len(experience)

for epoch in range(epochs):
    predicted_salary = beta0 + beta1 * experience
    error = salary - predicted_salary
    dbeta0 = (-2/n) * np.sum(error) #intercept
    dbeta1 = (-2/n) *np.sum(experience * error) #slope
    beta0 = beta0 - learning_rate * dbeta0
    beta1 = beta1 - learning_rate * dbeta1

print("Intercept:", beta0)
print("Slope:", beta1)

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay, accuracy_score, precision_score, recall_score, f1_score, roc_curve, roc_auc_score, classification_report)
```
